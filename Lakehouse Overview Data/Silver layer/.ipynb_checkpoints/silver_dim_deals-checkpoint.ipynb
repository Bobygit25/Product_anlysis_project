```python
df_deals['status'] = df_deals['status'].str.lower()
```

```python
df_deals['currency'].unique()
```

```python
date_columns = ['closed_at', 'created_at']  # to convert date columns from object to date
df_deals[date_columns] = df_deals[date_columns].apply(pd.to_datetime)
```

```python
df_deals['days_cycle_days'] = (df_deals['closed_at'] - df_deals['created_at']).dt.days
```

```python
df_deals['days_last_seen'] = (pd.Timestamp.now() - df_deals['closed_at']).dt.days
```

```python
df_deals['created_at'] = df_deals['created_at'].dt.date
df_deals['closed_at'] = df_deals['closed_at'].dt.date
```

```python
df_deals = df_deals.drop('days_cycle_days', axis = 1)
```
